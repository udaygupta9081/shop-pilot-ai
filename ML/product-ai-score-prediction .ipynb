```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder,OrdinalEncoder,LabelEncoder,StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
```

```python
tr1 = ColumnTransformer(transformers=[
    (
        'OneHotEncoding',
        OneHotEncoder(
            sparse_output=True,
            handle_unknown='ignore'
        ),
        ['categoryName']
    ),
    (
        'StandardScaling',
        StandardScaler(),
        ['stars', 'reviews', 'price', 'boughtInLastMonth']
    )
])
```

```python
tr1
```

```python
pipe1 = Pipeline(steps=[
    ('preprocessing', tr1),
    ('model', RandomForestClassifier(
        n_estimators=100,
        max_depth=15,
        min_samples_split=10,
        min_samples_leaf=4,
        max_features='sqrt',
        max_leaf_nodes=100,
        min_impurity_decrease=0.01,
        bootstrap=True,
        max_samples=0.8,
        class_weight='balanced',
        random_state=42,
        n_jobs=-1,
        warm_start=True
    ))
])
```

```python
X_fit = X_train.iloc[:30_000]

tr1.fit(X_fit)

print("✅ Preprocessor fitted successfully")

del X_fit

import gc
gc.collect()
```

```python
X_test_batch = X_train.iloc[:100]

X_test_transformed = tr1.transform(X_test_batch)

print("✅ Transform successful")
print("Shape:", X_test_transformed.shape)

del X_test_batch, X_test_transformed
gc.collect()
```

```python
model = RandomForestClassifier(
    n_estimators=10,
    max_depth=15,
    min_samples_split=10,
    min_samples_leaf=4,
    max_features='sqrt',
    max_leaf_nodes=100,
    min_impurity_decrease=0.01,
    bootstrap=True,
    max_samples=0.8,
    class_weight='balanced',
    random_state=42,
    n_jobs=-1,
    warm_start=True
)

batch_size = 30_000
trees_per_batch = 5

for start in range(0, len(X_train), batch_size):

    end = min(start + batch_size, len(X_train))

    print(f"\nProcessing rows {start:,} → {end:,}")

    X_batch = X_train.iloc[start:end]
    y_batch = y_train.iloc[start:end]

    X_batch_transformed = tr1.transform(X_batch)

    model.n_estimators += trees_per_batch

    model.fit(X_batch_transformed, y_batch)

    print(f"✅ Completed | Trees: {model.n_estimators}")

    del X_batch
    del y_batch
    del X_batch_transformed

    gc.collect()
```

```python
X_test_transformed = tr1.transform(X_test)
y_pred = model.predict(X_test_transformed)
```

```python
print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

```python
cm = confusion_matrix(y_test, y_pred)

print(cm)
```

```python
model = RandomForestClassifier(
    n_estimators=5,
    max_depth=15,
    min_samples_split=10,
    min_samples_leaf=4,
    max_features='sqrt',
    max_leaf_nodes=100,
    min_impurity_decrease=0.01,
    bootstrap=True,
    max_samples=0.8,
    class_weight='balanced',
    random_state=42,
    n_jobs=-1,
    warm_start=True
)

batch_size = 30_000
trees_per_batch = 5

for start in range(0, len(X_train), batch_size):

    end = min(start + batch_size, len(X_train))

    print(f"\nProcessing rows {start:,} → {end:,}")

    # Get batch
    X_batch = X_train.iloc[start:end]
    y_batch = y_train.iloc[start:end]

    # Transform batch
    X_batch_transformed = tr1.transform(X_batch)

    # Add new trees
    if start != 0:
        model.n_estimators += trees_per_batch

    # Train
    model.fit(
        X_batch_transformed,
        y_batch
    )

    print(f"✅ Completed | Trees: {model.n_estimators}")

    # Free memory
    del X_batch
    del y_batch
    del X_batch_transformed

    gc.collect()
```

```python
X_test_transformed = tr1.transform(X_test)
y_pred = model.predict(X_test_transformed)
print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
cm = confusion_matrix(y_test, y_pred)
print(cm)
```

```python
from sklearn.linear_model import SGDClassifier
```

```python
from sklearn.utils.class_weight import compute_class_weight
classes = np.unique(y_train)

# class_weights = {         This is the first hyperparameter where i get the best accuracy THIRD
#     0: 0.5,
#     1: 50
# }

# Now this is the best accuracy model  FIRST
class_weights = {         
    0: 1,
    1: 20
}

# class_weights = {         This is the second
#     0: 1,
#     1: 30
# }
# class_weights = {
#     0: 1,
#     1: 50
# }

model = SGDClassifier(
    loss='hinge',
    alpha=0.0001,
    max_iter=1000,
    tol=1e-3,
    random_state=42
)

batch_size = 30_000

for start in range(0, len(X_train), batch_size):

    end = min(start + batch_size, len(X_train))

    X_batch = X_train.iloc[start:end]
    y_batch = y_train.iloc[start:end]

    X_batch_transformed = tr1.transform(X_batch)

    sample_weights = np.array([
        class_weights[label]
        for label in y_batch
    ])

    model.partial_fit(
        X_batch_transformed,
        y_batch,
        classes=classes,
        sample_weight=sample_weights
    )

    print(f"✅ Completed: {start:,} → {end:,}")

    del X_batch
    del y_batch
    del X_batch_transformed
    del sample_weights

    gc.collect()
```

```python
X_test_transformed = tr1.transform(X_test)
y_pred = model.predict(X_test_transformed)
print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
cm = confusion_matrix(y_test, y_pred)
print(cm)
```
