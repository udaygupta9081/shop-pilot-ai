```python

import numpy as np
import gc

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.linear_model import SGDClassifier
from sklearn.pipeline import Pipeline
from sklearn.base import BaseEstimator, TransformerMixin
from scipy import sparse


# 1. Preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        (
            'OneHotEncoding',
            OneHotEncoder(
                sparse_output=True,
                handle_unknown='ignore'
            ),
            ['categoryName']
        ),
        (
            'StandardScaling',
            StandardScaler(),
            ['stars', 'reviews', 'price', 'boughtInLastMonth']
        )
    ]
)


# 2. Fit the preprocessor
# Keep your existing approach of fitting on 30,000 rows
X_fit = X_train.iloc[:30_000]

preprocessor.fit(X_fit)

del X_fit
gc.collect()


# 3. Define feature weights
feature_weights = {
    'stars': 2.0,
    'reviews': 2.0,
    'price': 2.0,
    'boughtInLastMonth': 3.0,
    'categoryName': 1.0
}

feature_names = preprocessor.get_feature_names_out()

weights = np.ones(len(feature_names))

for i, name in enumerate(feature_names):
    for feature, weight in feature_weights.items():
        if feature in name:
            weights[i] = weight


# 4. Custom transformer for feature weighting
class FeatureWeighter(BaseEstimator, TransformerMixin):

    def __init__(self, weights):
        self.weights = weights

    def fit(self, X, y=None):
        return self

    def transform(self, X):
        if sparse.issparse(X):
            return X.multiply(self.weights).tocsr()

        return X * self.weights


feature_weighter = FeatureWeighter(weights)


# 5. Classifier
class_weights = {
    0: 1,
    1: 20
}

classifier = SGDClassifier(
    loss='log_loss',
    alpha=0.0001,
    max_iter=1000,
    tol=1e-3,
    random_state=42
)


# 6. Create one complete pipeline
pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('feature_weighter', feature_weighter),
    ('classifier', classifier)
])


# 7. Incremental batch training
classes = np.unique(y_train)
batch_size = 30_000

for start in range(0, len(X_train), batch_size):

    end = min(start + batch_size, len(X_train))

    X_batch = X_train.iloc[start:end]
    y_batch = y_train.iloc[start:end]

    # Pipeline preprocessing
    X_transformed = pipeline.named_steps[
        'preprocessor'
    ].transform(X_batch)

    # Pipeline feature weighting
    X_transformed = pipeline.named_steps[
        'feature_weighter'
    ].transform(X_transformed)

    # Sample weights for class imbalance
    sample_weights = y_batch.map(class_weights).to_numpy()

    # Incremental training
    pipeline.named_steps['classifier'].partial_fit(
        X_transformed,
        y_batch,
        classes=classes,
        sample_weight=sample_weights
    )

    print(f"Completed: {start:,} → {end:,}")

    del X_batch, y_batch, X_transformed, sample_weights
    gc.collect()


print("Pipeline training completed successfully!")
```

```python

import numpy as np
import pandas as pd

def calculate_product_score(df):
    result = df.copy()

    # Configurable importance weights; must sum to 1
    score_weights = {
        'rating': 0.35,
        'popularity': 0.30,
        'reviews': 0.20,
        'affordability': 0.15
    }

    # Normalize rating: 0 to 100
    rating_score = (
        result['stars'].clip(0, 5).fillna(0) / 5
    ) * 100

    # Log scaling prevents huge counts from dominating
    popularity_score = (
        np.log1p(result['boughtInLastMonth'].clip(lower=0))
        / np.log1p(
            max(result['boughtInLastMonth'].max(), 1)
        )
    ) * 100

    reviews_score = (
        np.log1p(result['reviews'].clip(lower=0))
        / np.log1p(max(result['reviews'].max(), 1))
    ) * 100

    # Lower price scores higher for affordability.
    # Relative to the products being scored.
    price = result['price'].clip(lower=0)
    max_price = max(price.max(), 1)

    affordability_score = (
        1 - price / max_price
    ) * 100

    result['AI_Score'] = (
        score_weights['rating'] * rating_score
        + score_weights['popularity'] * popularity_score
        + score_weights['reviews'] * reviews_score
        + score_weights['affordability'] * affordability_score
    ).round(2)

    return result


scored_products = calculate_product_score(X_test)

print(
    scored_products.sort_values(
        'AI_Score', ascending=False
    ).head(20)
)
```

# Final Pipeline of the model

```python

import numpy as np
import pandas as pd
import gc
import joblib

from scipy import sparse

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.linear_model import SGDClassifier
from sklearn.pipeline import Pipeline
from sklearn.base import BaseEstimator, TransformerMixin

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    roc_auc_score,
    average_precision_score
)


# =====================================================
# 1. CONFIGURATION
# =====================================================

BATCH_SIZE = 30_000
CLASS_WEIGHTS = {0: 1, 1: 50}
PREDICTION_THRESHOLD = 0.50

# Customize feature importance for the ML classifier
FEATURE_WEIGHTS = {
    'stars': 6.0,
    'reviews': 4.5,
    'price': 1.0,
    'boughtInLastMonth': 7.0,
    'categoryName': 1.0
}

# Customize weights for the independent product score
PRODUCT_SCORE_WEIGHTS = {
    'rating': 0.4,
    'popularity': 0.5,
    'reviews': 0.45,
    'affordability': 0.15
}

# Fixed reference values for consistent scores across batches.
# Adjust these to fit the scale and units of your dataset.
POPULARITY_REFERENCE = 5000
REVIEWS_REFERENCE = 10000
PRICE_REFERENCE = 50000


# =====================================================
# 2. PREPROCESSING
# =====================================================

preprocessor = ColumnTransformer(
    transformers=[
        (
            'OneHotEncoding',
            OneHotEncoder(
                sparse_output=True,
                handle_unknown='ignore'
            ),
            ['categoryName']
        ),
        (
            'StandardScaling',
            StandardScaler(),
            [
                'stars',
                'reviews',
                'price',
                'boughtInLastMonth'
            ]
        )
    ]
)

# Fit on the complete training data so the encoder
# learns all categories present in the training set.
preprocessor.fit(X_train)

print("Preprocessor fitted successfully")


# =====================================================
# 3. FEATURE WEIGHTING TRANSFORMER
# =====================================================

feature_names = preprocessor.get_feature_names_out()
feature_multipliers = np.ones(len(feature_names))

for i, name in enumerate(feature_names):
    for feature, weight in FEATURE_WEIGHTS.items():
        if feature in name:
            feature_multipliers[i] = weight


class FeatureWeighter(BaseEstimator, TransformerMixin):

    def __init__(self, weights):
        self.weights = weights

    def fit(self, X, y=None):
        return self

    def transform(self, X):
        if sparse.issparse(X):
            return X.multiply(self.weights).tocsr()

        return X * self.weights


feature_weighter = FeatureWeighter(feature_multipliers)


# =====================================================
# 4. CLASSIFIER
# =====================================================

classifier = SGDClassifier(
    loss='log_loss',
    alpha=0.0001,
    max_iter=1000,
    tol=1e-3,
    random_state=42,
    class_weight=CLASS_WEIGHTS
)


# =====================================================
# 5. COMPLETE ML PIPELINE
# =====================================================

pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('feature_weighter', feature_weighter),
    ('classifier', classifier)
])


# =====================================================
# 6. INCREMENTAL BATCH TRAINING
# =====================================================

classes = np.unique(y_train)

for start in range(0, len(X_train), BATCH_SIZE):

    end = min(start + BATCH_SIZE, len(X_train))

    X_batch = X_train.iloc[start:end]
    y_batch = y_train.iloc[start:end]

    # Transform raw data
    X_transformed = pipeline.named_steps[
        'preprocessor'
    ].transform(X_batch)

    # Apply user-defined feature multipliers
    X_transformed = pipeline.named_steps[
        'feature_weighter'
    ].transform(X_transformed)

    # Class weights are applied inside SGDClassifier
    pipeline.named_steps['classifier'].partial_fit(
        X_transformed,
        y_batch,
        classes=classes
    )

    print(f"Trained: {start:,} -> {end:,}")

    del X_batch, y_batch, X_transformed
    gc.collect()

print("ML pipeline training completed")


# =====================================================
# 7. PRODUCT SCORE CALCULATOR
# =====================================================

def calculate_product_score(df):

    result = df.copy()

    stars = pd.to_numeric(
        result['stars'], errors='coerce'
    ).fillna(0).clip(0, 5)

    reviews = pd.to_numeric(
        result['reviews'], errors='coerce'
    ).fillna(0).clip(lower=0)

    purchases = pd.to_numeric(
        result['boughtInLastMonth'], errors='coerce'
    ).fillna(0).clip(lower=0)

    price = pd.to_numeric(
        result['price'], errors='coerce'
    ).fillna(PRICE_REFERENCE).clip(lower=0)

    # Rating score: 0-100
    rating_score = stars / 5 * 100

    # Log scaling limits domination by extremely high counts
    popularity_score = (
        np.log1p(purchases)
        / np.log1p(POPULARITY_REFERENCE)
        * 100
    ).clip(0, 100)

    review_score = (
        np.log1p(reviews)
        / np.log1p(REVIEWS_REFERENCE)
        * 100
    ).clip(0, 100)

    # Affordability: lower price receives a higher score.
    # This assumes cheaper products are preferable.
    affordability_score = (
        1 - price / PRICE_REFERENCE
    ).clip(0, 1) * 100

    # Combine your chosen weights
    result['AI_Product_Score'] = (
        PRODUCT_SCORE_WEIGHTS['rating'] * rating_score
        + PRODUCT_SCORE_WEIGHTS['popularity'] * popularity_score
        + PRODUCT_SCORE_WEIGHTS['reviews'] * review_score
        + PRODUCT_SCORE_WEIGHTS['affordability']
        * affordability_score
    ).round(2)

    return result


# =====================================================
# 8. PREDICTION FUNCTION
# =====================================================

def predict_products(raw_products):

    # Raw products go directly into the ML pipeline
    probabilities = pipeline.predict_proba(raw_products)

    positive_index = np.where(
        pipeline.named_steps['classifier'].classes_ == 1
    )[0][0]

    # Model-estimated probability of bestseller class
    bestseller_probability = probabilities[:, positive_index]

    # Separate heuristic product-quality score
    results = calculate_product_score(raw_products)

    results['Bestseller_Probability'] = (
        bestseller_probability * 100
    ).round(2)

    results['Predicted_Bestseller'] = (
        bestseller_probability >= PREDICTION_THRESHOLD
    ).astype(int)

    return results


# =====================================================
# 9. EVALUATE ON HELD-OUT TEST DATA
# =====================================================

test_results = predict_products(X_test)

y_pred = test_results['Predicted_Bestseller'].to_numpy()
y_prob = (
    test_results['Bestseller_Probability'].to_numpy() / 100
)

print("\nAccuracy:",
      accuracy_score(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(
    y_test, y_pred, zero_division=0
))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nROC-AUC:",
      roc_auc_score(y_test, y_prob))

print("\nAverage Precision:",
      average_precision_score(y_test, y_prob))


# =====================================================
# 10. SAVE THE RESULTS AND MODEL
# =====================================================

test_results.to_csv(
    'shop_pilot_scored_products.csv',
    index=False
)

joblib.dump(
    {
        'pipeline': pipeline,
        'product_score_weights': PRODUCT_SCORE_WEIGHTS,
        'prediction_threshold': PREDICTION_THRESHOLD,
        'popularity_reference': POPULARITY_REFERENCE,
        'reviews_reference': REVIEWS_REFERENCE,
        'price_reference': PRICE_REFERENCE
    },
    'shop_pilot_ai_pipeline.joblib'
)

print("\nSaved product scores and trained pipeline")
```

```python
import joblib
import numpy as np
import pandas as pd

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_auc_score,
    average_precision_score
)

# ============================================================
# 1. LOAD SAVED SHOPPILOT AI MODEL
# ============================================================

saved = joblib.load("shop_pilot_ai_pipeline.joblib")

model = saved["pipeline"]
PRODUCT_SCORE_WEIGHTS = saved["product_score_weights"]
PREDICTION_THRESHOLD = saved["prediction_threshold"]

POPULARITY_REFERENCE = saved["popularity_reference"]
REVIEWS_REFERENCE = saved["reviews_reference"]
PRICE_REFERENCE = saved["price_reference"]

print("ShopPilot AI loaded successfully!")
print("Prediction threshold:", PREDICTION_THRESHOLD)
print("Product score weights:", PRODUCT_SCORE_WEIGHTS)


# ============================================================
# 2. PREPARE FULL DATASET
# ============================================================

# Replace df with the variable containing your full dataset.
feature_cols = [
    "categoryName",
    "stars",
    "reviews",
    "price",
    "boughtInLastMonth"
]

X = df[feature_cols].copy()

# Keep original labels for evaluation, if available.
has_labels = "isBestSeller" in df.columns

if has_labels:
    y = df["isBestSeller"].astype(int).to_numpy()


# ============================================================
# 3. CALCULATE AI PRODUCT SCORE (0-100)
# ============================================================

def calculate_ai_product_score(products):
    products = products.copy()

    # Handle missing or invalid numerical values safely.
    stars = pd.to_numeric(products["stars"], errors="coerce").fillna(0).clip(0, 5)
    reviews = pd.to_numeric(products["reviews"], errors="coerce").fillna(0).clip(lower=0)
    price = pd.to_numeric(products["price"], errors="coerce").fillna(PRICE_REFERENCE).clip(lower=0)
    bought = pd.to_numeric(
        products["boughtInLastMonth"], errors="coerce"
    ).fillna(0).clip(lower=0)

    rating_score = stars / 5 * 100

    popularity_score = (
        np.log1p(bought) / np.log1p(POPULARITY_REFERENCE) * 100
    ).clip(0, 100)

    review_score = (
        np.log1p(reviews) / np.log1p(REVIEWS_REFERENCE) * 100
    ).clip(0, 100)

    affordability_score = (
        1 - price / PRICE_REFERENCE
    ).clip(0, 1) * 100

    component_scores = {
        "stars": rating_score,
        "boughtInLastMonth": popularity_score,
        "reviews": review_score,
        "price": affordability_score
    }

    total_weight = sum(
        PRODUCT_SCORE_WEIGHTS.get(feature, 0)
        for feature in component_scores
    )

    if total_weight <= 0:
        raise ValueError("No valid product-score weights found.")

    score = sum(
        component_scores[feature] * PRODUCT_SCORE_WEIGHTS.get(feature, 0)
        for feature in component_scores
    ) / total_weight

    # Round and cap strictly below 100.
    score = np.asarray(score, dtype=float)
    score = np.nan_to_num(score, nan=0.0, posinf=99.99, neginf=0.0)
    score = np.clip(score, 0, 99.99)

    return np.round(score, 2)


# ============================================================
# 4. PREDICT THE FULL DATASET IN BATCHES
# ============================================================

batch_size = 50_000
probability_batches = []

positive_index = list(model.classes_).index(1)

for start in range(0, len(X), batch_size):
    batch = X.iloc[start:start + batch_size]

    batch_probabilities = model.predict_proba(batch)
    probability_batches.append(batch_probabilities[:, positive_index])

    print(f"Processed {min(start + batch_size, len(X)):,} / {len(X):,}")

bestseller_probabilities = np.concatenate(probability_batches)

predicted_labels = (
    bestseller_probabilities >= PREDICTION_THRESHOLD
).astype(int)


# ============================================================
# 5. BUILD FULL RESULTS
# ============================================================

results = df.copy()

results["Bestseller_Probability_Percent"] = np.round(
    bestseller_probabilities * 100, 2
)

results["Predicted_Bestseller"] = predicted_labels

results["Prediction_Label"] = np.where(
    predicted_labels == 1,
    "Bestseller",
    "Not Bestseller"
)

results["AI_Product_Score"] = calculate_ai_product_score(X)

# Ensure the displayed score remains strictly below 100.
results["AI_Product_Score"] = results["AI_Product_Score"].clip(upper=99.99)


# ============================================================
# 6. EVALUATE IF ACTUAL LABELS EXIST
# ============================================================

if has_labels:
    print("\n========== FULL DATA EVALUATION ==========")
    print("Total products:", f"{len(X):,}")
    print("Actual bestsellers:", f"{y.sum():,}")
    print("Predicted bestsellers:", f"{predicted_labels.sum():,}")

    print("\nAccuracy :", round(accuracy_score(y, predicted_labels), 4))
    print("Precision:", round(precision_score(y, predicted_labels, zero_division=0), 4))
    print("Recall   :", round(recall_score(y, predicted_labels, zero_division=0), 4))
    print("F1-score :", round(f1_score(y, predicted_labels, zero_division=0), 4))

    print("\nConfusion Matrix:")
    print(confusion_matrix(y, predicted_labels))

    print("\nClassification Report:")
    print(classification_report(
        y,
        predicted_labels,
        target_names=["Not Bestseller", "Bestseller"],
        zero_division=0
    ))

    print("ROC-AUC:", round(roc_auc_score(y, bestseller_probabilities), 4))
    print("Average Precision:", round(
        average_precision_score(y, bestseller_probabilities), 4
    ))


# ============================================================
# 7. DISPLAY RANDOM PRODUCTS
# ============================================================

sample_size = min(20, len(results))

random_products = results.sample(
    n=sample_size,
    random_state=None
)

display_cols = [
    "categoryName",
    "stars",
    "reviews",
    "price",
    "boughtInLastMonth",
    "AI_Product_Score",
    "Bestseller_Probability_Percent",
    "Prediction_Label"
]

if has_labels:
    display_cols.insert(1, "isBestSeller")

print("\n========== RANDOM SHOPPILOT AI PRODUCTS ==========")

display(
    random_products[display_cols]
    .sort_values("AI_Product_Score", ascending=False)
    .reset_index(drop=True)
)


# ============================================================
# 8. SAVE ALL PREDICTIONS
# ============================================================

results.to_csv("Predictions.csv", index=False)

print("\nSaved: shop_pilot_ai_full_predictions.csv")
print("Total products scored:", f"{len(results):,}")
```
